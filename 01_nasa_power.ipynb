```python
import requests, pandas as pd

def get_power(lat, lon, start="20240101", end="20241231"):
    url = "https://power.larc.nasa.gov/api/temporal/daily/point"
    params = {
        "parameters": "PRECTOTCORR,ALLSKY_SFC_SW_DWN",  # rainfall, sunlight
        "community": "RE", "latitude": lat, "longitude": lon,
        "start": start, "end": end, "format": "JSON",
    }
    data = requests.get(url, params=params, timeout=60).json()
    df = pd.DataFrame(data["properties"]["parameter"])
    df.index = pd.to_datetime(df.index)
    df = df.replace(-999, pd.NA)  # NASA uses -999 for missing values
    return df.rename(columns={"PRECTOTCORR": "rain_mm",
                              "ALLSKY_SFC_SW_DWN": "irradiance_kwh_m2"})

jodhpur = get_power(26.24, 73.02)
print(jodhpur.head())
print(jodhpur.describe())
```
